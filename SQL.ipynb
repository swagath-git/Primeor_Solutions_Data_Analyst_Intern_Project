```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('ecommerce.db')
```

```python
df = pd.read_csv('cleaned_ecommerce_data.csv')
df
```

```python
df.to_sql('cleaned_ecommerce_data', conn, if_exists='replace', index=False)
```

```python
pd.read_sql_query("SELECT * FROM cleaned_ecommerce_data LIMIT 5", conn)
```

```python
# TOP 10 PROFOTABLE PRODUCTS
query = """
SELECT product_name, SUM(profit) AS total_profit
FROM cleaned_ecommerce_data
GROUP BY product_name
ORDER BY total_profit DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
# TOP 10 CUSTOMERS BY SALES
query = """
SELECT customer_name, SUM(sales) AS total_sales
FROM cleaned_ecommerce_data
GROUP BY customer_name
ORDER BY total_sales DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
# REGION WISE TOTAL SALES
query = """
SELECT region, SUM(sales) AS total_sales
FROM cleaned_ecommerce_data
GROUP BY region
ORDER BY total_sales DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# CATEGORY WISE AVERAGE PROFIT
query = """
SELECT category, AVG(profit) AS average_profit
FROM cleaned_ecommerce_data
GROUP BY category;
"""

pd.read_sql_query(query, conn)
```

```python
# HIGHEST DISCOUNT CATEGORY
query = """
SELECT category, AVG(discount) AS average_discount
FROM cleaned_ecommerce_data
GROUP BY category
ORDER BY average_discount DESC
LIMIT 1;
"""

pd.read_sql_query(query, conn)
```

```python
# OREDRS WITH NEGATIVE PROFIT
query = """
SELECT *
FROM cleaned_ecommerce_data
WHERE profit < 0;
"""

pd.read_sql_query(query, conn)
```

```python
# MONTHLY SALES TREND
query = """
SELECT
    strftime('%Y', order_date) AS year,
    strftime('%m', order_date) AS month,
    SUM(sales) AS total_sales
FROM cleaned_ecommerce_data
GROUP BY year, month
ORDER BY year, month;
"""

pd.read_sql_query(query, conn)
```

```python
# MARKE-WISE REVENUE ANALYSIS
query = """
SELECT market, SUM(sales) AS revenue
FROM cleaned_ecommerce_data
GROUP BY market
ORDER BY revenue DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# TOP PERFORMING SUB CATEGORIES
query = """
SELECT sub_category, SUM(sales) AS total_sales
FROM cleaned_ecommerce_data
GROUP BY sub_category
ORDER BY total_sales DESC;
"""

pd.read_sql_query(query, conn)
```

```python
# SHIP MODE USAGE ANALYSIS
query = """
SELECT ship_mode, COUNT(*) AS usage_count
FROM cleaned_ecommerce_data
GROUP BY ship_mode
ORDER BY usage_count DESC;
"""

pd.read_sql_query(query, conn)
```
