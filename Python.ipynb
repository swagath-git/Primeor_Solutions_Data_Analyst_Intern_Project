## **Handling missing/Null values**

```python
# Customer name is in text column, we replace the missing values with "unknown" 

df['customer_name'] = df['customer_name'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
# profit column is a numerical column, so filling with 0

df.dropna(subset=['profit'], inplace=True)
```

```python
df.isnull().sum()
```

## **3. Standardize Inconsistent Text Values**

```python
# Checking Text columns here

txt_columns = df.select_dtypes(include='object').columns
txt_columns
```

```python
# Verifing the text columns here

for col in txt_columns:
    print("\n", col)
    print(df[col].unique())
```

```python
# Re-checking the columns and confirming

for col in txt_columns:
    values = df[col].dropna().astype(str)
    print(col, ":", (values != values.str.strip()).sum())
```
